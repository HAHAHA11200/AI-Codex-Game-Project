```python
class Roman():
    def __init__(self, name, email, schedule, idNum):
        self.name = name
        self.email = email
        self.schedule = schedule
        self.id = idNum

    def __str__(self):
        return f"Name: {self.name}, Email: {self.email}"

    def print_schedule(self):
        for key, value in self.schedule.items():
            print(f"{key} Block: {value}")
```

```python
class USStudent(Roman):
    def __init__(self, name, email, gpa, grade, schedule, idNum):
        super().__init__(name, email, schedule, idNum)
        self.gpa = gpa
        self.grade = grade

    def __str__(self):
        usroman_str = super().__str__()
        return f"{usroman_str}, GPA: {self.gpa}, Email{self.email}"

    def advance_grade(self):
        if self.grade < 12:
            self.grade += 1

    def generate_email(self):
        first,last = self.name.split()
        print(first[0]+last[:7:]+"@lsoc.org")
```
