```python
class Roman():
    def __init__(self, name, schedule, idNum):
        self.name = name
        self.schedule = schedule
        self.id = idNum

    def __str__(self):
        return f"Name: {self.name}, ID Number: {self.id}"

    def print_schedule(self):
        for key, value in self.schedule.items():
            print(f"{key} Block: {value}")
```

```python
# USStudent class EXTENDS the Roman class by including it as a parameter
class USStudent(Roman):
    def __init__(self, name, gpa, grade, schedule, idNum):
      # Super is a metheod that allows inheritance from the poarent
        super().__init__(name, schedule, idNum)
      # Include the attributes that are exclusive to this object
        self.gpa = gpa
        self.grade = grade

    def __str__(self):
      # This defines the str for JUST the student
        usroman_str = super().__str__()
        return f"{usroman_str}, GPA: {self.gpa}, Grade: {self.grade}"

    # You must include the methods for the specific object
    def advance_grade(self):
        if self.grade < 12:
            self.grade += 1
        return self.grade

    def generate_email(self):
        first,last = self.name.split()
        print(first[0]+last[:7:]+"@lsoc.org")
```

```python
class Teach(Roman):
  def __init__(self, name, subject, schedule, idNum, experience):
    super().__init__(name, schedule, idNum)
    self.sub = subject
    self.exp = experience
  def __str__(self):
    teach_str = super().__str__()
    return f"{teach_str}, Experience: {self.exp} Years, Subject: {self.sub}"
  def generate_email(self):
    first,last = self.name.split()
    print(first[0]+last[:7:]+"@latinschool.org")
  def checkfree(self):
    for block,thing in self.schedule.items():
      if thing == 'Free':
        return f'{self.name} is free during {block} Block.'
```

```python
USStudent1 = USStudent('Jake Goldstein', 2.0, 9, {'A':'CSP2','B':'CSP2','C':'Spanish','D':'CSP2','E':'Hon. Physics','F':'Wellness','G':'Hon. Algebra 2'}, 201676)
Teach1 = Teach('Metro Boomin', 'Modern Music Production', {'A':'Music Production','B':'Free','C':'Band','D':'Rythm','E':'Band','F':'Free','G':'Free'}, 389560, 15)
print(USStudent1)
print(Teach1)
USStudent1.print_schedule()
USStudent1.advance_grade()
print(USStudent1)
USStudent1.generate_email()
Teach1.generate_email()
Teach1.checkfree()
```
