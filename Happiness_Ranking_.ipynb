For this project I set out to see what aspects of a country contribute to how happy the people of that country feel. My dataset is a collection of data from the Gallup World Poll to rank countries happiness.

The question I would like to pase is "what makes a county's population happy." My hypothesis is that the happiness score of a country will matter less about material or nation wide things like GDP or the government, but will more likely be dependant on things that the individual will feel more like how much social support they get or how free they feel to make choices.

```python
import matplotlib.pyplot as plt  # Just importing resources
import numpy as np # from w3schools
import csv

with open('2019.csv','r') as f:        #Opening the CSV
  reader = csv.reader(f)
  headings = next(reader)
  headings.pop(0)
  rows = []
  for row in reader:              #Separating the Rows By
    rows.append(row)
  country = []
  for row in rows:
    country.append(row[1])          #country
  rank = []
  for row in rows:
    rank.append(float(row[0]))       #rank
  score = []
  for row in rows:
    score.append(float(row[2]))     #score
  gdp = []
  for row in rows:
    gdp.append(float(row[3]))       #GDP
  support = []
  for row in rows:
    support.append(float(row[4]))  #support
  health = []
  for row in rows:
    health.append(float(row[5]))   #health
  freedom = []
  for row in rows:
    freedom.append(float(row[6]))  #freedom
  generosity = []
  for row in rows:
    generosity.append(float(row[7]))   #generosity
  corruption = []
  for row in rows:
    corruption.append(float(row[8])) # and corruption
  x = np.array(gdp)
  y = np.array(score)
  plt.xlabel('GDP Per Capita')
  plt.ylabel('Happiness score')
  plt.scatter(x,y)    #basic first graph comparing GDP per capita and the hapiness score
  plt.show
```

This graph shows the relationship between country's GDP and their happiness score using a scatter plot. The graph is mostly a direct correlation but there are many outliers and a lot of the points deviate from the center of the graph. This sort of supports my point that GDP is not the largest decider in happiness.
